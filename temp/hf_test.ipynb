A very simple test notebook for loading the HuggingFace dataset and displaying random instances.

```python
from datasets import load_dataset
import random
from IPython.display import display
```

```python
ds = load_dataset('gemixin/touch-fl', split='train')
print(ds)
```

```python
# Get a random sample from the dataset
sample = ds[random.randint(0, len(ds) - 1)]
# Print the metadata
print({key: value for key, value in sample.items() if key != 'image'})
# Display the image
display(sample['image'])
```
