## Google Colab Script Runner Notebook

Run the experiment scripts in Google Colab and save results and checkpoints to Google Drive.
Select a GPU runtime, then run the setup cells in order.

The repository is cloned from GitHub, so push local pipeline changes before starting a new session.

````python
# Initial imports
import sys
import os
import subprocess
from google.colab import drive
````

````python
# Clone the repository to the Colab environment if it is not already present
repo_path = "/content/touch-ex"
if not os.path.isdir(repo_path):
    subprocess.run(
        ["git", "clone", "https://github.com/gemixin/touch-ex.git", repo_path],
        check=True,
    )
else:
    print(f"Using existing repository: {repo_path}")
````

````python
# Add the repository to the Python path and use it as the working directory
# Now we can import modules from the cloned repository
if repo_path not in sys.path:
    sys.path.insert(0, repo_path)
os.chdir(repo_path)
````

````python
# Mount Google Drive so experiment outputs persist after the Colab session ends
drive.mount("/content/drive")

# Set up directories for experiment outputs inside Google Drive
OUTPUT_DIR = "/content/drive/MyDrive/Colab Notebooks/touch-ex"
RESULTS_DIR = os.path.join(OUTPUT_DIR, "results")
CHECKPOINT_DIR = os.path.join(OUTPUT_DIR, "checkpoints")
os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(CHECKPOINT_DIR, exist_ok=True)
````
