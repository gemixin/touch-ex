## Exploring Force Data
This notebook explores the Touch-Ex dataset numerical force data statistics.

```python
# Allow imports from the parent directory
import sys
import os

module_path = os.path.abspath(os.path.join(".."))
sys.path.append(module_path)
```

```python
# Imports
import data.utils as utils
```

```python
# Load the dataset into DataFrames
dfs = utils.load_touch_ex_dataset()
```

```python
# Extract the training and validation_unseen_matched DataFrames from the dictionary
train_df = dfs["train"]
validation_unseen_matched_df = dfs["validation_unseen_matched"]
```

```python
# Get column names (same for both train_df and validation_unseen_matched_df)
train_df.columns
```

### FSR Voltage

```python
train_df.groupby("force_level").agg({"fsr_voltage": ["mean", "std", "min", "max"]})
```

```python
validation_unseen_matched_df.groupby("force_level").agg({"fsr_voltage": ["mean", "std", "min", "max"]})
```

### Force (Newtons)

```python
train_df.groupby("force_level").agg({"force_n": ["mean", "std", "min", "max"]})
```

```python
validation_unseen_matched_df.groupby("force_level").agg({"force_n": ["mean", "std", "min", "max"]})
```
