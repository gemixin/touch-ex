## Exploring Force Data
This notebook explores the Touch-Ex dataset numerical force data statistics.

```python
# Imports
from datasets import load_dataset
from IPython.display import display, Markdown
```

### Load Dataset from HuggingFace

```python
splits = [
    "train",
    "validation_unseen_matched",
    "test_unseen_matched",
    "test_unseen_related",
]
dfs = {split: load_dataset("gemixin/touch-ex", split=split).to_pandas() for split in splits}
dfs.keys()
```

### FSR Voltage

```python
for split, df in dfs.items():
    display(Markdown(f"**{split}:**"))
    display(df.groupby("force_level").agg({"fsr_voltage": ["mean", "std", "min", "max"]}))
    print("")
```

### Force (Newtons)

```python
for split, df in dfs.items():
    display(Markdown(f"**{split}:**"))
    display(df.groupby("force_level").agg({"force_n": ["mean", "std", "min", "max"]}))
    print("")
```
