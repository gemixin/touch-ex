## View Experiment Results

This notebook allows you to view an experiments Parquet file as a DataFrame.


```python
# Imports
import pandas as pd
from IPython.display import display
```

```python
# Set the results file path
PARQUET_PATH = "../results/object_classify/final_evaluation/experiments.parquet"
```

```python
experiments_df = pd.read_parquet(PARQUET_PATH)
with pd.option_context("display.max_columns", None):
    display(experiments_df)
```
